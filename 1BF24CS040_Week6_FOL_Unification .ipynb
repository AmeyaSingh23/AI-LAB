```python
def is_variable(expr):
    return isinstance(expr, str) and expr.islower() and len(expr) == 1

def is_constant(expr):
    return isinstance(expr, str) and (expr.isupper() or expr[0].isupper()) and not isinstance(expr, list)

def apply_substitution(sub, expr):
    if isinstance(expr, list):
        return [apply_substitution(sub, item) for item in expr]
    if expr in sub:
        return sub[expr]
    return expr

def unify(expr1, expr2, theta=None):
    if theta is None:
        theta = {}

    expr1 = apply_substitution(theta, expr1)
    expr2 = apply_substitution(theta, expr2)

    if expr1 == expr2:
        return theta

    if is_variable(expr1):
        if occur_check(expr1, expr2):
            return None
        theta[expr1] = expr2
        return theta

    if is_variable(expr2):
        if occur_check(expr2, expr1):
            return None
        theta[expr2] = expr1
        return theta

    if isinstance(expr1, list) and isinstance(expr2, list):
        if len(expr1) != len(expr2):
            return None
        for item1, item2 in zip(expr1, expr2):
            theta = unify(item1, item2, theta)
            if theta is None:
                return None
        return theta

    return None

def occur_check(var, expr):
    """Returns True if variable occurs in the expression."""
    if var == expr:
        return True
    if isinstance(expr, list):
        return any(occur_check(var, item) for item in expr)
    return False

def print_unification(e1, e2):
    print(f"Expression 1: {e1}")
    print(f"Expression 2: {e2}")
    result = unify(e1, e2)
    if result is None:
        print("Result      : Unification Failed!")
    else:
        print(f"MGU (Theta) : {result}")
    print("-" * 40)
```
