```python
import random

def display_board(board):
    print("\n" + " | ".join(board[0]))
    print("---------")
    print(" | ".join(board[1]))
    print("---------")
    print(" | ".join(board[2]) + "\n")

def is_valid_move(board, row, col):
    return 0 <= row < 3 and 0 <= col < 3 and board[row][col] == ' '

def get_player_move(board):
    while True:
        try:
            move = input("Enter your move (row, column, e.g., 1,2): ")
            row, col = map(int, move.split(','))
            row -= 1  # 0-indexed
            col -= 1  # 0-indexed
            if is_valid_move(board, row, col):
                return row, col
            else:
                print("Invalid move. That spot is already taken or out of bounds. Try again.")
        except ValueError:
            print("Invalid input. Please enter row and column as numbers separated by a comma (e.g., 1,2).")

def check_win(board, player):
    for row in board:
        if all([cell == player for cell in row]):
            return True
    for col_idx in range(3):
        if all([board[row_idx][col_idx] == player for row_idx in range(3)]):
            return True
    if (board[0][0] == player and board[1][1] == player and board[2][2] == player) or \
       (board[0][2] == player and board[1][1] == player and board[2][0] == player):
        return True
    return False

def check_draw(board):
    for row in board:
        for cell in row:
            if cell == ' ':
                return False
    return True

def get_computer_move(board, computer_symbol, player_symbol):
    for r in range(3):
        for c in range(3):
            if is_valid_move(board, r, c):
                board[r][c] = computer_symbol
                if check_win(board, computer_symbol):
                    return r, c
                board[r][c] = ' '

    for r in range(3):
        for c in range(3):
            if is_valid_move(board, r, c):
                board[r][c] = player_symbol
                if check_win(board, player_symbol):
                    board[r][c] = ' '
                    return r, c
                board[r][c] = ' '
    if is_valid_move(board, 1, 1):
        return 1, 1

    corners = [(0, 0), (0, 2), (2, 0), (2, 2)]
    random.shuffle(corners)
    for r, c in corners:
        if is_valid_move(board, r, c):
            return r, c

    sides = [(0, 1), (1, 0), (1, 2), (2, 1)]
    random.shuffle(sides)
    for r, c in sides:
        if is_valid_move(board, r, c):
            return r, c

    empty_cells = []
    for r in range(3):
        for c in range(3):
            if board[r][c] == ' ':
                empty_cells.append((r, c))
    return random.choice(empty_cells)

def play_game():
    board = [[' ' for _ in range(3)] for _ in range(3)]
    player_symbol = 'X'
    computer_symbol = 'O'
    current_turn = 'player'

    print("Welcome to Tic Tac Toe!")
    print("You are 'X', the computer is 'O'.")
    display_board(board)

    while True:
        if current_turn == 'player':
            print("Your turn (", player_symbol, ")")
            row, col = get_player_move(board)
            board[row][col] = player_symbol
        else:
            print("Computer's turn (", computer_symbol, ")")
            row, col = get_computer_move(board, computer_symbol, player_symbol)
            board[row][col] = computer_symbol
            print(f"Computer plays at {row + 1},{col + 1}")

        display_board(board)

        if check_win(board, player_symbol):
            print("Congratulations! You won!")
            break
        elif check_win(board, computer_symbol):
            print("Computer wins! Better luck next time.")
            break
        elif check_draw(board):
            print("It's a draw!")
            break

        current_turn = 'computer' if current_turn == 'player' else 'player'

play_game()
```

```python
import random
import time
import os

class RoomEnvironment:
    def __init__(self, num_rows, num_cols, initial_dirty_percentage=0.5):
        self.num_rows = num_rows
        self.num_cols = num_cols
        self.grid = [['dirty' if random.random() < initial_dirty_percentage else 'clean' for _ in range(num_cols)] for _ in range(num_rows)]
        self.agent_pos = (0, 0) # Agent starts at top-left

    def is_dirty(self, row, col):
        return self.grid[row][col] == 'dirty'

    def clean_room(self, row, col):
        if self.is_dirty(row, col):
            self.grid[row][col] = 'clean'
            return True
        return False

    def get_percept(self, row, col):
        return self.grid[row][col]

    def display(self):
        os.system('cls' if os.name == 'nt' else 'clear') # Clear console for animation
        print("\n--- Room Environment ---")
        for r_idx, row in enumerate(self.grid):
            for c_idx, cell in enumerate(row):
                if (r_idx, c_idx) == self.agent_pos:
                    print(f"[{'V' if cell == 'clean' else 'D'}]", end=" ") # V for Vac, D for Dirty with Vac
                else:
                    print(f" {cell[0].upper()} ", end=" ") # C for Clean, D for Dirty
            print()
        print("------------------------")

    def all_rooms_clean(self):
        for row in self.grid:
            for cell in row:
                if cell == 'dirty':
                    return False
        return True

class VacuumCleanerAgent:
    def __init__(self, environment):
        self.environment = environment
        self.row, self.col = environment.agent_pos
        self.cleaned_count = 0
        self.path = [(self.row, self.col)] # To keep track of movements

    def choose_action(self):
        current_percept = self.environment.get_percept(self.row, self.col)

        if current_percept == 'dirty':
            return 'clean', None # Return two values, None for target_pos
        else:
            # Explore: try to find a dirty room or a new path
            possible_moves = []
            # Check neighbors for dirty rooms first
            for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                new_row, new_col = self.row + dr, self.col + dc
                if 0 <= new_row < self.environment.num_rows and 0 <= new_col < self.environment.num_cols:
                    if self.environment.is_dirty(new_row, new_col):
                        return 'move', (new_row, new_col)
                    possible_moves.append((new_row, new_col))

            # If no dirty neighbors, move to an unvisited valid neighbor or randomly
            if possible_moves:
                unvisited_moves = [m for m in possible_moves if m not in self.path[-min(len(self.path), 5):]] # Avoid immediate backtracking
                if unvisited_moves:
                    return 'move', random.choice(unvisited_moves)
                else:
                    return 'move', random.choice(possible_moves) # If stuck, move randomly
            else:
                return 'idle', None # Return two values, None for target_pos

    def execute_action(self, action, target_pos=None):
        if action == 'clean':
            if self.environment.clean_room(self.row, self.col):
                self.cleaned_count += 1
                # print(f"Agent cleaned room at ({self.row+1},{self.col+1})")
        elif action == 'move' and target_pos:
            self.row, self.col = target_pos
            self.environment.agent_pos = (self.row, self.col)
            self.path.append((self.row, self.col))
            # print(f"Agent moved to ({self.row+1},{self.col+1})")
        # elif action == 'idle':
            # print("Agent is idle.")

def run_simulation(num_rows, num_cols, steps_limit=100, initial_dirty_percentage=0.5, delay=0.2):
    env = RoomEnvironment(num_rows, num_cols, initial_dirty_percentage)
    agent = VacuumCleanerAgent(env)

    print("Initial Environment:")
    env.display()
    time.sleep(1)

    for step in range(steps_limit):
        if env.all_rooms_clean():
            print(f"\nAll rooms cleaned in {step} steps!")
            break

        action_type, target_pos = agent.choose_action()
        agent.execute_action(action_type, target_pos)

        env.display()
        time.sleep(delay)
    else:
        print(f"\nSimulation finished after {steps_limit} steps.")

    if env.all_rooms_clean():
        print("Final State: All rooms clean!")
    else:
        print("Final State: Some rooms might still be dirty.")
    print(f"Total rooms cleaned by agent: {agent.cleaned_count}")


run_simulation(num_rows=3, num_cols=3, initial_dirty_percentage=0.7, steps_limit=50, delay=0.3)
```
