```python
class Symbol:
    """Represents a logical variable."""
    def __init__(self, name):
        self.name = name
    def __repr__(self):
        return self.name

# Helper functions to build logical expression strings
def Not(op):
    return f"(not {op})"

def Or(left, right):
    return f"({left} or {right})"

def Implies(left, right):
    return f"((not {left}) or {right})"
```

```python
def tt_entails(KB, alpha):
    """Determine if the Knowledge Base (KB) entails query (alpha)."""
    # Collect all unique variable names
    symbols = sorted(list({'P', 'Q', 'R'}))

    print(f"Symbols involved in KB & Alpha: {symbols}\n")
    print(f"{' | '.join(symbols)} | KB | Alpha | Status")
    print("-" * (len(symbols)*4 + 25))

    return tt_check_all(KB, alpha, symbols, {})

def tt_check_all(KB, alpha, symbols, model):
    if not symbols:
        # Evaluate using Python's built-in eval by passing the model dictionary as namespace
        kb_true = all(eval(str(sentence), {}, model) for sentence in KB)
        alpha_true = eval(str(alpha), {}, model)

        vals = [str(model[sym])[0] for sym in sorted(model.keys())]
        status = "Valid" if (not kb_true or alpha_true) else "Fails (Counter-example!)"
        print(f" {' | '.join(vals)} |  {str(kb_true)[0]} |   {str(alpha_true)[0]}   | {status}")

        return alpha_true if kb_true else True
    else:
        P, rest = symbols[0], symbols[1:]
        return tt_check_all(KB, alpha, rest, {**model, P: True}) and \
               tt_check_all(KB, alpha, rest, {**model, P: False})
```

```python
P = Symbol('P')
Q = Symbol('Q')

KB1 = [
    Implies(P, Q),
    P
]
Alpha1 = Q

print("Evaluating Test Case 1: Modus Ponens (P => Q, P) |= Q")
result1 = tt_entails(KB1, Alpha1)
print(f"\nDoes KB entail Alpha? {result1}\n")
```

```python
R = Symbol('R')
KB2 = [
    Or(P, Q),
    Not(P)
]
Alpha2 = R

print("Evaluating Test Case 2: (P | Q, ~P) |= R")
result2 = tt_entails(KB2, Alpha2)
print(f"\nDoes KB entail Alpha? {result2}\n")
```
